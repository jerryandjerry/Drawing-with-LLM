```python
def competition_score(vqa_score: float, ocr_score: float, aesthetic_score: float) -> float:
    """
    Computes the final competition score given VQA score, OCR score, and aesthetic score.

    Follows the official logic:
      - Harmonic mean of VQA and Aesthetic scores with beta=0.5
      - Multiplied by OCR score

    Parameters
    ----------
    vqa_score : float
        The VQA task accuracy score (0 to 1).

    ocr_score : float
        The OCR penalty score (0 to 1).

    aesthetic_score : float
        The aesthetic quality score (0 to 1).

    Returns
    -------
    float
        Final competition score.
    """
    beta = 0.5

    # Handle zero edge cases
    if vqa_score <= 0 or aesthetic_score <= 0:
        return 0.0

    harmonic = (1 + beta**2) * (vqa_score * aesthetic_score) / (beta**2 * vqa_score + aesthetic_score)

    final_score = harmonic * ocr_score

    return final_score
```

```python
vqa = 0.701379
ocr = 0.9
aes = 0.529669

score = competition_score(vqa, ocr, aes)
print(f"Competition score: {score:.4f}")
```
