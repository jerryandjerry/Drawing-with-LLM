```python
from vllm import LLM, SamplingParams
from transformers import AutoTokenizer

# Few-shot prompt with examples
FEW_SHOT_PROMPT_TEMPLATE = """<bos> [INST] Generate a "are there" question based on the norms in the given caption.

Caption: A dog running through a grassy field.
Question: Are there a dog and a grassy field?

Caption: A group of people standing around a table with food.
Question: Are there people, a table and food?

Caption: A city skyline with tall buildings at sunset.
Question: Are there city skyline, tall buildings and sunset?

Caption: {caption}
Question: [/INST]"""

class PromptQuestionGenerator:
    def __init__(self, llm, tokenizer):
        self.llm = llm
        self.tokenizer = tokenizer

    def generate(self, caption: str, max_new_tokens=64) -> str:
        prompt = FEW_SHOT_PROMPT_TEMPLATE.format(caption=caption)

        sampling_params = SamplingParams(
            max_tokens=max_new_tokens,
            temperature=0.8,
            top_p=0.95,
            top_k=64,
            repetition_penalty=1.1
        )

        outputs = self.llm.generate(prompt, sampling_params)
        return outputs[0].outputs[0].text.strip()

# --- Model loading globally ---
model_path = "gemma-3-1b-it-bnb-4bit"
llm = LLM(model=model_path, dtype="float16", gpu_memory_utilization=0.9)
tokenizer = AutoTokenizer.from_pretrained(model_path)

# Instantiate the generator
generator = PromptQuestionGenerator(llm, tokenizer)


```

```python
# Example test
caption = "A red sports car driving along a mountain road."
question = generator.generate(caption)
print("Generated Question:", question)
```
